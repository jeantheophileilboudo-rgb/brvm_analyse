```python
import pandas as pd

fichier = pd.read_excel('../data/raw/BRVM_Etats_Financiers_1998_2025.xlsx', sheet_name=None)
print(list(fichier.keys()))
```

```python
bilan = fichier['Bilan']
resultat = fichier['Compte de résultat']
flux = fichier['Flux de trésorerie']

print('Bilan :', bilan.shape)
print('Compte de résultat :', resultat.shape)
print('Flux de trésorerie :', flux.shape)
```

```python
print('BILAN :', bilan.columns.tolist())
print()
print('COMPTE DE RÉSULTAT :', resultat.columns.tolist())
print()
print('FLUX DE TRÉSORERIE :', flux.columns.tolist())
```

```python
print('Nombre de sociétés :', bilan['Société'].nunique())
print('Première année :', bilan['Exercice'].min())
print('Dernière année :', bilan['Exercice'].max())
print('Lignes en double :', bilan.duplicated(subset=['Société', 'Exercice']).sum())
print()
print(bilan['Exercice'].value_counts().sort_index())
```

```python
pd.set_option('display.max_rows', 100)

annees_par_societe = bilan.groupby('Société')['Exercice'].agg(['count', 'min', 'max'])
annees_par_societe.sort_values('count')
```

```python
manquantes = []
for societe, groupe in bilan.groupby('Société'):
    annees = set(groupe['Exercice'])
    attendues = set(range(min(annees), max(annees) + 1))
    for annee in sorted(attendues - annees):
        manquantes.append((societe, annee))

pd.DataFrame(manquantes, columns=['Société', 'Année manquante'])
```

```python
def cles(feuille):
    return set(zip(feuille['Société'], feuille['Exercice']))

k_bilan = cles(bilan)
k_resultat = cles(resultat)
k_flux = cles(flux)

print('Dans le bilan mais pas dans le compte de résultat :', len(k_bilan - k_resultat))
print('Dans le compte de résultat mais pas dans le bilan :', len(k_resultat - k_bilan))
print('Dans le bilan mais pas dans les flux :', len(k_bilan - k_flux))
print('Dans les flux mais pas dans le bilan :', len(k_flux - k_bilan))
print()
print('Doublons compte de résultat :', resultat.duplicated(subset=['Société', 'Exercice']).sum())
print('Doublons flux :', flux.duplicated(subset=['Société', 'Exercice']).sum())
print()
print('Noms différents (bilan / résultat) :', set(bilan['Société']) ^ set(resultat['Société']))
print('Noms différents (bilan / flux) :', set(bilan['Société']) ^ set(flux['Société']))
```

```python
sans_flux = pd.DataFrame(sorted(k_bilan - k_flux), columns=['Société', 'Exercice'])
print(sans_flux.groupby('Société').size().sort_values(ascending=False))
print()
print('Dans le compte de résultat mais pas dans le bilan :')
print(sorted(k_resultat - k_bilan))
```

```python
sans_flux = pd.DataFrame(sorted(k_bilan - k_flux), columns=['Société', 'Exercice'])
print(sans_flux.groupby('Société').size().sort_values(ascending=False))
print()
print('Dans le compte de résultat mais pas dans le bilan :')
print(sorted(k_resultat - k_bilan))
```

```python
colonnes_banque = ['Depots de la clientele',
                   'Creance sur la clientele (Prêts)',
                   'Creance interbancaire']

est_financiere = (bilan[colonnes_banque].fillna(0) != 0).any(axis=1).groupby(bilan['Société']).any()

financieres = sorted(est_financiere[est_financiere].index)
autres = sorted(est_financiere[~est_financiere].index)

print('FINANCIÈRES :', len(financieres))
print(financieres)
print()
print('AUTRES :', len(autres))
print(autres)
```

```python
type_societe = {s: 'Financière' for s in financieres}
type_societe.update({s: 'Autre' for s in autres})

for feuille in (bilan, resultat, flux):
    feuille['Type'] = feuille['Société'].map(type_societe)

print(bilan['Type'].value_counts())
print()
print(resultat['Type'].value_counts())
print()
print(flux['Type'].value_counts())
print()
print('Lignes sans type :', bilan['Type'].isna().sum(), resultat['Type'].isna().sum(), flux['Type'].isna().sum())
```

```python
bilan['Ecart'] = bilan['TOTAL ACTIF'] - bilan['TOTAL PASSIF ET CAPITAUX PROPRES']
bilan['Ecart_relatif'] = bilan['Ecart'].abs() / bilan['TOTAL ACTIF'].abs()

print('Total actif manquant :', bilan['TOTAL ACTIF'].isna().sum())
print('Total passif et capitaux propres manquant :', bilan['TOTAL PASSIF ET CAPITAUX PROPRES'].isna().sum())
print('Lignes comparables :', bilan['Ecart_relatif'].notna().sum())
print('Écart supérieur à 0,1 % :', (bilan['Ecart_relatif'] > 0.001).sum())
print('Écart supérieur à 1 % :', (bilan['Ecart_relatif'] > 0.01).sum())
print()

colonnes = ['Société', 'Exercice', 'Type', 'TOTAL ACTIF', 'TOTAL PASSIF ET CAPITAUX PROPRES', 'Ecart_relatif']
bilan.loc[bilan['Ecart_relatif'] > 0.01, colonnes].sort_values('Ecart_relatif', ascending=False).head(15)
```
