```python
def cles(feuille):
    return set(zip(feuille['Société'], feuille['Exercice']))

k_bilan = cles(bilan)
k_resultat = cles(resultat)
k_flux = cles(flux)

print('Dans le bilan mais pas dans le compte de résultat :', len(k_bilan - k_resultat))
print('Dans le compte de résultat mais pas dans le bilan :', len(k_resultat - k_bilan))
print('Dans le bilan mais pas dans les flux :', len(k_bilan - k_flux))
print('Dans les flux mais pas dans le bilan :', len(k_flux - k_bilan))
print()
print('Doublons compte de résultat :', resultat.duplicated(subset=['Société', 'Exercice']).sum())
print('Doublons flux :', flux.duplicated(subset=['Société', 'Exercice']).sum())
print()
print('Noms différents (bilan / résultat) :', set(bilan['Société']) ^ set(resultat['Société']))
print('Noms différents (bilan / flux) :', set(bilan['Société']) ^ set(flux['Société']))
```

```python
colonnes_banque = ['Depots de la clientele',
                   'Creance sur la clientele (Prêts)',
                   'Creance interbancaire']

est_financiere = (bilan[colonnes_banque].fillna(0) != 0).any(axis=1).groupby(bilan['Société']).any()

financieres = sorted(est_financiere[est_financiere].index)
autres = sorted(est_financiere[~est_financiere].index)

print('FINANCIÈRES :', len(financieres))
print(financieres)
print()
print('AUTRES :', len(autres))
print(autres)
```

```python
bilan['Ecart'] = bilan['TOTAL ACTIF'] - bilan['TOTAL PASSIF ET CAPITAUX PROPRES']
bilan['Ecart_relatif'] = bilan['Ecart'].abs() / bilan['TOTAL ACTIF'].abs()

print('Total actif manquant :', bilan['TOTAL ACTIF'].isna().sum())
print('Total passif et capitaux propres manquant :', bilan['TOTAL PASSIF ET CAPITAUX PROPRES'].isna().sum())
print('Lignes comparables :', bilan['Ecart_relatif'].notna().sum())
print('Écart supérieur à 0,1 % :', (bilan['Ecart_relatif'] > 0.001).sum())
print('Écart supérieur à 1 % :', (bilan['Ecart_relatif'] > 0.01).sum())
print()

colonnes = ['Société', 'Exercice', 'Type', 'TOTAL ACTIF', 'TOTAL PASSIF ET CAPITAUX PROPRES', 'Ecart_relatif']
bilan.loc[bilan['Ecart_relatif'] > 0.01, colonnes].sort_values('Ecart_relatif', ascending=False).head(15)
```

```python
manquants = bilan[bilan['TOTAL ACTIF'].isna()][['Société', 'Exercice', 'Type']]
print('Lignes sans total actif :', len(manquants))
print(manquants.sort_values(['Société', 'Exercice']).to_string(index=False))
print()

autres_bilan = bilan[bilan['Type'] == 'Autre'].copy()
somme = autres_bilan['TOTAL ACTIF NON COURANT'] + autres_bilan['TOTAL ACTIF COURANT']
ecart2 = (somme - autres_bilan['TOTAL ACTIF']).abs() / autres_bilan['TOTAL ACTIF'].abs()
print('Sociétés non financières comparables :', ecart2.notna().sum())
print('Écart supérieur à 1 % :', (ecart2 > 0.01).sum())

colonnes = ['Société', 'Exercice', 'TOTAL ACTIF NON COURANT', 'TOTAL ACTIF COURANT', 'TOTAL ACTIF']
autres_bilan.loc[ecart2 > 0.01, colonnes].head(15)
```

```python
import os

cles = ['Société', 'Exercice', 'Type']
bilan_propre = bilan.drop(columns=['Ecart', 'Ecart_relatif'], errors='ignore')

base = bilan_propre.merge(resultat, on=cles, how='outer').merge(flux, on=cles, how='outer')
base = base.sort_values(['Société', 'Exercice']).reset_index(drop=True)

print('Lignes :', base.shape[0], '| Colonnes :', base.shape[1])
print('Pas de doublon société-année :', not base.duplicated(subset=['Société', 'Exercice']).any())

os.makedirs('../data/processed', exist_ok=True)
base.to_csv('../data/processed/base_brvm.csv', index=False, encoding='utf-8-sig')
```

```python
import numpy as np

def ratio(num, den):
    r = num / den.where(den > 0)
    return r.replace([np.inf, -np.inf], np.nan)

base = base.sort_values(['Société', 'Exercice']).reset_index(drop=True)

cp = base['Capitaux propres (interet + part minoritaire)']
ca = base["Chiffre d'affaires"]

# Ratios communs
base['ROE'] = ratio(base['RESULTAT NET'], cp)
base['ROA'] = ratio(base['RESULTAT NET'], base['TOTAL ACTIF'])

# Sociétés non financières
base['Marge nette'] = ratio(base['RESULTAT NET'], ca)
base['Marge exploitation'] = ratio(base["RESULTAT D'EXPLOITATION"], ca)
dettes = base[['Dette financiere a long terme', 'Dette financiere a court terme']].sum(axis=1, min_count=1)
base['Endettement'] = ratio(dettes, cp)
base['Liquidite'] = ratio(base['TOTAL ACTIF COURANT'], base['TOTAL PASSIF COURANT'])

ca_prec = base.groupby('Société')["Chiffre d'affaires"].shift(1)
annee_prec = base.groupby('Société')['Exercice'].shift(1)
croissance = ca / ca_prec.where(ca_prec > 0) - 1
base['Croissance CA'] = croissance.where(base['Exercice'] - annee_prec == 1)

base['Flux libre'] = base['Net cash des operations'] + base["Cash d'acquisition d'immobilisation"]
base['Conversion cash'] = ratio(base['Net cash des operations'], base['RESULTAT NET'])

# Banques
pnb = base['Produit Net Bancaire (PNB)']
base['Coefficient exploitation'] = 1 - ratio(base["RESULTAT BRUT D'EXPLOITATION (RBE)"], pnb)
base['Cout du risque / PNB'] = ratio(base['Cout du risque (Provisions impayés)'].abs(), pnb)
base['Credits / Depots'] = ratio(base['Creance sur la clientele (Prêts)'], base['Depots de la clientele'])

colonnes_autres = ['Marge nette', 'Marge exploitation', 'ROE', 'ROA', 'Endettement',
                   'Liquidite', 'Croissance CA', 'Conversion cash']
colonnes_banques = ['ROE', 'ROA', 'Coefficient exploitation', 'Cout du risque / PNB', 'Credits / Depots']

a = base[base['Type'] == 'Autre'][colonnes_autres]
b = base[base['Type'] == 'Financière'][colonnes_banques]

print('SOCIÉTÉS NON FINANCIÈRES')
print(pd.DataFrame({'Médiane': a.median().round(3), 'Nb valeurs': a.count()}))
print()
print('SOCIÉTÉS FINANCIÈRES')
print(pd.DataFrame({'Médiane': b.median().round(3), 'Nb valeurs': b.count()}))

base.to_csv('../data/processed/ratios_brvm.csv', index=False, encoding='utf-8-sig')
```

```python
def md(s):
    return (s / 1e9).round(2)

print("Chiffre d'affaires négatif :", (base["Chiffre d'affaires"] < 0).sum())
print("Chiffre d'affaires nul :", (base["Chiffre d'affaires"] == 0).sum())
print("Capitaux propres négatifs ou nuls :", (base['Capitaux propres (interet + part minoritaire)'] <= 0).sum())
print()

negatifs = base[base["Chiffre d'affaires"] < 0][['Société', 'Exercice', "Chiffre d'affaires"]]
print("Lignes avec chiffre d'affaires négatif :")
print(negatifs.to_string(index=False))
print()

for nom, debut, fin in [('NESTLE CI', 2010, 2020), ('SMB', 2012, 2017)]:
    g = base[(base['Société'] == nom) & base['Exercice'].between(debut, fin)]
    t = pd.DataFrame({'Exercice': g['Exercice'],
                      "Chiffre d'affaires": md(g["Chiffre d'affaires"]),
                      'Résultat net': md(g['RESULTAT NET']),
                      'Capitaux propres': md(g['Capitaux propres (interet + part minoritaire)']),
                      'Total actif': md(g['TOTAL ACTIF'])})
    print(nom, '(milliards de FCFA)')
    print(t.to_string(index=False))
    print()
```

```python
cp = base['Capitaux propres (interet + part minoritaire)']
actif = base['TOTAL ACTIF']

base['Alerte_CP_faible'] = cp <= 0.01 * actif
base['Alerte_CA_negatif'] = base["Chiffre d'affaires"] < 0
base['ROE_fiable'] = base['ROE'].where(~base['Alerte_CP_faible'])

print("Lignes avec capitaux propres inférieurs à 1 % de l'actif (ou négatifs) :", base['Alerte_CP_faible'].sum())
print()

resume = base.groupby('Type').agg(
    lignes_alerte=('Alerte_CP_faible', 'sum'),
    ROE_median_brut=('ROE', 'median'),
    ROE_median_fiable=('ROE_fiable', 'median'),
    ROE_max_brut=('ROE', 'max'),
    ROE_max_fiable=('ROE_fiable', 'max')).round(2)
print(resume)
print()

print('Sociétés les plus concernées :')
print(base[base['Alerte_CP_faible']].groupby('Société').size().sort_values(ascending=False).head(10))

nouvelles = [
    ("SETAO CI", "2005", "Compte de résultat", "Valeur impossible",
     "Chiffre d'affaires négatif (environ -16 millions de FCFA)", "À vérifier"),
    ("NESTLE CI", "2016", "Toutes les feuilles", "Ligne vide",
     "Société-année présente sans aucun montant au bilan ni au compte de résultat", "À vérifier"),
    ("NESTLE CI, SMB", "2012-2019", "Bilan", "Capitaux propres quasi nuls",
     "ROE extrême car capitaux propres très faibles : ratio non interprétable", "Documenté"),
]
journal = pd.concat([journal, pd.DataFrame(nouvelles, columns=journal.columns)], ignore_index=True)
journal.to_csv('../outputs/journal_anomalies.csv', index=False, encoding='utf-8-sig')
print()
print("Lignes du journal d'anomalies :", len(journal))

base.to_csv('../data/processed/ratios_brvm.csv', index=False, encoding='utf-8-sig')
```
